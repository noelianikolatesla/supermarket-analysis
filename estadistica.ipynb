1- Cargar datos

```python
import pandas as pd

df = pd.read_csv("data/SampleSuperstore.csv")
df.head()
```

2- Limpiar datos

```python
df.isnull().sum()
df.dropna(inplace=True)

print(df.columns)

```

3- Estadistica Descriptiva

```python
df.describe()


```

4- Grafico (1): Categorias de producto mas vendidas

```python
import matplotlib.pyplot as plt

df['Sales'].plot(kind='hist')
plt.title('Distribución del volumen de ventas')
plt.xlabel('Ventas')
plt.show()

```

Grafico (2): Ventas totales (histograma)

```python
df['Sales'].plot(kind='hist')
plt.title('Distribución del volumen de ventas')
plt.xlabel('Ventas')
plt.show()
```

Grafico (3): Beneficio total por categoria

```python
df.groupby('Category')['Profit'].sum().plot(kind='bar')
plt.title('Beneficio total por categoría')
plt.ylabel('Beneficio (€)')
plt.show()
```

Grafico (4): Cantidad vendida por subcategoria

```python
df['Sub-Category'].value_counts().plot(kind='bar')
plt.title('Cantidad de ventas por subcategoría')
plt.xlabel('Subcategoría')
plt.ylabel('Cantidad')
plt.show()
```

1. Beneficio promedio por categoría

```python
df.groupby('Category')['Profit'].mean()
```

2. Beneficio TOTAL por categoría (importante para negocio)

```python
df.groupby('Category')['Profit'].sum()
```

3. Gráfico del beneficio total por categoría

```python
df.groupby('Category')['Profit'].sum().plot(kind='bar')
plt.title('Beneficio total por categoría')
plt.ylabel('Beneficio (€)')
plt.show()
```

SECCION DE CONCLUSIONES

```python
# Beneficio total por categoría
beneficio_categoria = df.groupby('Category')['Profit'].sum()
print("Beneficio total por categoría:\n")
print(beneficio_categoria)

# Categoría más rentable
categoria_rentable = beneficio_categoria.idxmax()
# Categoría menos rentable (posibles pérdidas)
categoria_peor = beneficio_categoria.idxmin()

print("\nCategoría más rentable:", categoria_rentable)
print("Categoría menos rentable:", categoria_peor)

print("\nSECCIÓN DE CONCLUSIONES")
print(
    f"Después de analizar los datos del supermercado, se observa que la "
    f"categoría más rentable es **{categoria_rentable}**, con un beneficio total "
    f"de **{beneficio_categoria[categoria_rentable]:.2f} €**, mientras que la categoría "
    f"menos rentable es **{categoria_peor}**, con **{beneficio_categoria[categoria_peor]:.2f} €**."
)
print(
    "Estas conclusiones pueden ayudar a la toma de decisiones estratégicas "
    "para mejorar la rentabilidad del negocio."
)
```
